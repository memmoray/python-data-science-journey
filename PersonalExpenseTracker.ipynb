```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

columns = ["Date", "Category","Description","Amount"]
expenses=pd.DataFrame(columns=columns)

expenses=pd.DataFrame([
    ["2025-12-01", "Groceries", "Supermarket", 45.50],
    ["2025-12-02", "Transport", "Uber", 12.00],
    ["2025-12-03", "Food", "Lunch", 15.75]], columns=columns)
def add_expense(date, category, description, amount):
    global expenses
    new_expense = pd.DataFrame([[date, category, description, amount]], columns=expenses.columns)
    expenses = pd.concat([expenses, new_expense], ignore_index=True)

# Example
add_expense("2025-12-04", "Entertainment", "Movie", 10.00)
expenses
# Total spent
print("Total Spent: $", expenses["Amount"].sum())

# Spending by category
category_summary = expenses.groupby("Category")["Amount"].sum()
print(category_summary)

# Plot spending by category
category_summary.plot(kind='bar', title="Spending by Category")
plt.ylabel("Amount ($)")
plt.show()

# Save your expenses to a CSV file
expenses.to_csv("expenses.csv", index=False)

# Load from CSV
expenses = pd.read_csv("expenses.csv")
expenses
```
