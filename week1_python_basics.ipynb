```python
print("Hello, Modesta!")
name="Modesta"
city="Austin"
Income=2000
print(f"My name is {name} and I live in {city}.")
```

```python
def monthly_savings(income, expenses):
  return income - expenses

# Assuming sample expenses of 480
sample_expenses = 480
savings = monthly_savings(Income, sample_expenses)
print(f"Your monthly savings are: {savings}")
```

```python
import pandas as pd

data = {
    "Category": ["Rent", "Food", "Transportation", "Savings"],
    "Amount": [480, 300, 120, 200]
}
df=pd.DataFrame(data)
print(df)
```

```python
import matplotlib.pyplot as plt
plt.bar(df["Category"], df["Amount"])
plt.title("Monthly Spending")
plt.xlabel("Category")
plt.ylabel("Amount($)")
plt.show()
```
